# Rotation of Array
> - the rot90() function is used to rotate an array by 90 degrees in the plane specified by axes
> - Syntax:
    *  numpy.rot90(m,k=1,axes=(0,1))

```python
import numpy as np

x = np.array([[1,1,1],
              [2,2,2]])

y = np.array([[4,4,4],
              [3,3,3]])

result = np.concatenate([x, y], axis=0)

print(result)
```

# Spliting of array

```python
#split [0:3] , [3:5] ,[5:7],[7:]
x=np.arange(1,10)
print(x)
np.split(x,[3,5,7])
```
