# Computation on numpy arrays
* comutation on numpy  arrays can be very fast or it can be very slow
* the key to makingit fast is to use 

```python
import numpy as np
x=np.array([1,2,3,4])
y=np.array([4,5,6,5])
xy=np.add(x,y)
print(xy)
```

```python
x=np.array([1,2,3,4])
y=np.array([4,5,6,5])
xy=np.subtract(x,y)
print(xy)
```

```python
x=np.array([1,2,3,4])
y=np.array([4,5,6,5])
xy=np.multiply(x,y)
print(xy)
```

```python
x=np.array([1,2,3,4])
y=np.array([4,5,6,5])
xy=np.divide(x,y)
print(xy)
```

```python
x=np.array([1,2,3,-1,-4,3])
np.absolute(x)
np.exp(x)
np.exp2(x)
np.log(x)
np.log2(x)
```

# Advance Ufunction Features

```python
x=np.arange(2,20,2)
print(x)
z=np.empty(9)
print(z)
print(np.multiply(x,10,out=z))
```

# 2.Aggregations(reduce method)
* if you like to reduce an array with particular operation we can use the reduce method of any ufucn
* A reduce repeatetly appiles a given operation to the elememt of array untill only a single result remains

```python
x=np.arange(1,11)
print(x)
np.add.reduce(x)
```

```python
x=np.arange(1,11)
print(x)
np.add.accumulate(x)
```

```python
x=np.arange(1,11)
print(x)
np.cumsum(x)
```
