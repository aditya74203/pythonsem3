write a program to find the sum of n nums

```python
def sum(n):
    total=0
    for i in range(1,n+1):
        total+=i   
    return total
n=int(input("Enter a number: "))
print("The sum of first",n,"numbers is:",sum(n))
```

```python
#using lambda function
sum=lambda n: n*(n+1)/2
n=int(input("Enter a number: "))
print("The sum of first",n,"numbers is:",sum(n))
```

```python
def get_name(person):
    return person['name']

people = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Charlie', 'age': 35}
]

names = list(map(get_name, people))
print(names)
```

```python
def even(n):
    if n%2==0:
        return True
    else:
        return False
num=(1,2,3,4,5,6,7,8,9)
even_numbers = list(filter(even, num))
print(even_numbers)
```

```python
def first_position(students):
    name,marks=students
    if marks > 80:
        return name
    else:
        return None
students={'Alice': 85, 'Bob': 75, 'Charlie': 90, 'David': 60}
first_positions = list(map(first_position, students.items()))
print(first_positions)
```

higher order function


```python
names = ["Aaditya", "Rahul", "Aman"]
marks = [85, 75, 90]

for name, mark in zip(names, marks):
    if mark > 80:
        print(name)
```
